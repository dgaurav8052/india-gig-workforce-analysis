## Data Cleaning and Preparations


```python
import pandas as pd
```

```python
df = pd.read_csv("../data/original_dataset.csv")
```

```python
df["secondary_income_source"].value_counts(dropna=False)
```

```python
df["secondary_income_source"] = df["secondary_income_source"].fillna("No secondary income")
```

```python
df["secondary_income_source"].isna().sum()
```

```python
df["injury_compensation_received"].value_counts(dropna=False)
```

```python
df["injury_compensation_received"] = df["injury_compensation_received"].fillna(
    "Not applicable"
)
```

```python
df["injury_compensation_received"].isna().sum()
```

```python
df["onboarding_document_issue"].value_counts(dropna=False)
```

```python
df["onboarding_document_issue"] = df["onboarding_document_issue"].fillna(
    "No issue reported"
)
```

```python
df["onboarding_document_issue"].isna().sum()
```

```python
df["vehicle_type"].value_counts(dropna=False)
```

```python
df["vehicle_type"] = df["vehicle_type"].fillna("Unknown")
```

```python
df["vehicle_type"].isna().sum()
```

```python
df.isna().sum()[df.isna().sum() > 0]
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df["primary_complaint"].value_counts()
```

```python
df["primary_complaint"] = df["primary_complaint"].replace(
    "No none", "No complaint"
)
```

```python
df["primary_complaint"].value_counts()
```

```python
df.to_csv("../data/cleaned_gig_workers.csv", index=False)
```
