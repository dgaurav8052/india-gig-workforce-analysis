```python
import pandas as pd

df = pd.read_csv("../data/cleaned_gig_workers.csv")

df.shape
```

```python
income_by_platform = df.groupby("platform_type").agg(
    workers_count = ("worker_id", "count"),
    average_net_income = ("monthly_net_income_inr", "mean"),
    median_net_income = ("monthly_net_income_inr", "median")
)

income_by_platform = income_by_platform.sort_values(
    "median_net_income", ascending=False
)

income_by_platform
```

```python
hours_income = df.groupby("platform_type").agg(
    average_daily_hours = ("daily_hours_worked", "mean"),
    median_daily_hours = ("daily_hours_worked", "median"),
    average_net_income = ("monthly_net_income_inr", "mean"),
    median_net_income = ("monthly_net_income_inr", "median")
)

hours_income.sort_values("median_net_income", ascending=False)
```

```python
df["vehicle_cost_burden_pct"] = (
    (df["monthly_fuel_cost_inr"] + df["monthly_emi_burden_inr"])
    / df["monthly_gross_income_inr"]
) * 100
```

```python
cost_by_platform = df.groupby("platform_type").agg(
    worker_count=("worker_id", "count"),
    median_gross_income=("monthly_gross_income_inr", "median"),
    median_net_income=("monthly_net_income_inr", "median"),
    median_fuel_cost=("monthly_fuel_cost_inr", "median"),
    median_emi=("monthly_emi_burden_inr", "median"),
    median_cost_burden=("vehicle_cost_burden_pct", "median")
)

cost_by_platform.sort_values(
    "median_cost_burden", ascending=False
)
```

## Key Findings

### 1. Income Differences Across Platform Types
- Home services had the highest median monthly net income at ₹15,009.
- Bike taxi had the lowest median monthly net income at ₹5,372.
- Median income varied considerably across platform types, despite average daily working hours being very similar (approximately 9.2 hours).

### 2. Vehicle Cost Burden
- Bike taxi had the highest median vehicle cost burden at 38.2% of gross income, followed by ride-hailing at 36.5%.
- Home services had the lowest median vehicle cost burden at 11.5%.
- Higher vehicle cost burdens coincided with lower median net incomes across platform types, although this does not establish a causal relationship.

### 3. Social Security Coverage
- 55.3% of records reported no social security coverage.
- An additional 15.0% reported being unsure about their coverage.
- Only 3.9% reported having both PF and ESI coverage.

### 4. Injury Compensation
- Of the 12,755 records reporting an injury in the previous 12 months, 54.6% reported receiving no compensation.
- Platform one-time payments were the second most common outcome, accounting for 18.2% of injured-worker records.
